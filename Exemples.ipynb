```python
import subprocess

def render_dfa(automate, output_png="tmp.png"):
    fig = automate.show_diagram()
    fig.write("tmp.dot")  # écrit juste le source DOT, pas de rendu natif en mémoire
    subprocess.run([
        r"/opt/homebrew/bin/dot",
        "-Tpng", "tmp.dot", "-o", output_png
    ], check=True)
    return output_png
```
